## Time resolution, completeness and timezone
We do **not** guess the timezone. Evidence: the longitude implied by the average clear-sky curve. Solar noon at longitude Î» happens at about 12:00 âˆ’ Î»/15 hours UTC.

```python
# CELL 6 â€” Check Time Interval, Completeness and Timezone
time_difference = df["time"].diff().dropna()
print("Most common time intervals:")
print(time_difference.value_counts().head())

expected = pd.date_range(df["time"].min(), df["time"].max(), freq="h")
missing_hours = len(expected.difference(pd.DatetimeIndex(df["time"])))
print("\nExpected hourly rows:", len(expected), "| Actual rows:", len(df), "| Missing hours:", missing_hours)
assert missing_hours == 0, "Gaps found: lag features below assume a continuous hourly series."

# Timezone evidence: centre of mass of the average clear-sky daily curve (hour-beginning stamps)
clear_by_hour = df.groupby(df["time"].dt.hour)["Clear_Sky_Solar_Radiation"].mean()
centre_utc = ((clear_by_hour.index + 0.5) * clear_by_hour).sum() / clear_by_hour.sum()
implied_longitude = (12 - centre_utc) * 15
print("\nClear-sky curve centre of mass:", round(centre_utc, 2), "h (timestamp clock)")
print("Implied longitude if the clock is UTC:", round(implied_longitude, 1), "deg E")
print("Damascus longitude is about 36.3 deg E -> timestamps are consistent with UTC, NOT local solar time.")

# EXPLICIT ASSUMPTION used for household behaviour: Syria is UTC+3 (permanent since Oct 2022).
LOCAL_UTC_OFFSET = 3
print("\nAssumption: Local clock = UTC +", LOCAL_UTC_OFFSET, "h (household usage times are defined in local time).")
```

## Units
Per the NASA POWER hourly documentation, `ALLSKY_SFC_SW_DWN` and `CLRSKY_SFC_SW_DWN` are irradiance in **W/mÂ²**, `T2M` is **Â°C** and `PRECTOTCORR` is **mm/hour**. The dataset itself carries no unit metadata, so Cell 8 runs physical sanity checks on the value ranges (e.g. radiation must stay below about 1400 W/mÂ²). Please confirm the units in your original NASA request.

```python
# CELL 7 â€” Clean Numeric Data and Check Anomalies
numeric_columns = ["Solar_Radiation", "Clear_Sky_Solar_Radiation", "Temperature", "Rain"]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# NASA-style missing value flag
for col in numeric_columns:
    df.loc[df[col] <= -900, col] = np.nan

print("Missing values:")
print(df[numeric_columns].isnull().sum())

print("\n--- Anomaly checks ---")
print("Negative solar values:", int((df["Solar_Radiation"] < 0).sum()), "/", int((df["Clear_Sky_Solar_Radiation"] < 0).sum()))
print("Negative rain values :", int((df["Rain"] < 0).sum()))
print("Max ALLSKY / CLRSKY  :", df["Solar_Radiation"].max(), "/", df["Clear_Sky_Solar_Radiation"].max(), "W/m2 (physical limit ~1400)")
print("Temperature range    :", df["Temperature"].min(), "to", df["Temperature"].max(), "C")
print("Max hourly rain      :", df["Rain"].max(), "mm/h")
print("ALLSKY > CLRSKY rows :", int((df["Solar_Radiation"] > df["Clear_Sky_Solar_Radiation"]).sum()),
      "| by more than 20 W/m2:", int((df["Solar_Radiation"] > df["Clear_Sky_Solar_Radiation"] + 20).sum()),
      "(small excursions are possible, e.g. cloud-edge reflection)")

for col in ["Solar_Radiation", "Clear_Sky_Solar_Radiation", "Rain"]:
    df[col] = df[col].clip(lower=0)
```

```python
# CELL 8 â€” Remove Incomplete Rows
df = df.dropna(subset=numeric_columns).reset_index(drop=True)
print("Rows after cleaning:", len(df))
```

```python
# CELL 9 â€” Create Time Features
df["Year"] = df["time"].dt.year
df["Month"] = df["time"].dt.month
df["Day"] = df["time"].dt.day
df["Hour"] = df["time"].dt.hour                  # UTC hour
df["DayOfWeek"] = df["time"].dt.dayofweek
df["DayOfYear"] = df["time"].dt.dayofyear

# Local clock (assumption: UTC+3), used for household behaviour and scheduling
df["Local_time"] = df["time"] + pd.Timedelta(hours=LOCAL_UTC_OFFSET)
df["Local_Hour"] = df["Local_time"].dt.hour
df["Local_Date"] = df["Local_time"].dt.date

df[["time", "Local_time", "Year", "Month", "Day", "Hour", "Local_Hour", "DayOfWeek", "DayOfYear"]].head()
```

```python
# CELL 10 â€” Create Solar Features
df["Is_Daytime"] = df["Clear_Sky_Solar_Radiation"] > 20     # sun clearly above the horizon

# Solar ratio (clearness): only defined when the clear-sky value is meaningful (avoids division by ~0)
df["Solar_Ratio"] = np.where(df["Is_Daytime"],
                             df["Solar_Radiation"] / df["Clear_Sky_Solar_Radiation"].replace(0, np.nan),
                             np.nan)
df["Solar_Ratio"] = df["Solar_Ratio"].clip(0, 1.5)

df["Solar_Difference"] = df["Clear_Sky_Solar_Radiation"] - df["Solar_Radiation"]

print("Daytime hours:", int(df["Is_Daytime"].sum()), "of", len(df))
display(df.loc[df["Is_Daytime"], ["Solar_Ratio", "Solar_Difference"]].describe())
```

# Part 3 â€” Solar availability classes (LOW / MEDIUM / HIGH)
Thresholds come from the real data: the 33rd and 66th percentiles of **daytime** all-sky radiation (hourly class) and of **daily** solar energy (daily class). Night hours are LOW by definition. The classes describe *how much solar power is available*, not only how clear the sky is, so winter days are naturally lower than summer days.

```python
# CELL 19 â€” Solar Availability Classes
day_values = df.loc[df["Is_Daytime"], "Solar_Radiation"]
HOURLY_P33, HOURLY_P66 = day_values.quantile([0.33, 0.66]).values
DAILY_P33, DAILY_P66 = daily["Solar_kWh_m2"].quantile([0.33, 0.66]).values

print("Hourly thresholds (daytime ALLSKY, W/mÂ²): LOW <", round(HOURLY_P33, 1),
      "| MEDIUM <", round(HOURLY_P66, 1), "| HIGH >=", round(HOURLY_P66, 1))
print("Daily thresholds (kWh/mÂ²/day):             LOW <", round(DAILY_P33, 2),
      "| MEDIUM <", round(DAILY_P66, 2), "| HIGH >=", round(DAILY_P66, 2))

def classify_hourly(x):
    x = np.asarray(x, dtype=float)
    return np.select([x < HOURLY_P33, x < HOURLY_P66], ["LOW", "MEDIUM"], default="HIGH")

def classify_daily(x):
    x = np.asarray(x, dtype=float)
    return np.select([x < DAILY_P33, x < DAILY_P66], ["LOW", "MEDIUM"], default="HIGH")

df["Solar_Class"] = classify_hourly(df["Solar_Radiation"])
daily["Daily_Class"] = classify_daily(daily["Solar_kWh_m2"])

print("\nHourly classes (daytime):")
print(df.loc[df["Is_Daytime"], "Solar_Class"].value_counts())
print("\nDaily classes:")
print(daily["Daily_Class"].value_counts())

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].hist(day_values, bins=50)
ax[0].axvline(HOURLY_P33, color="r", ls="--")
ax[0].axvline(HOURLY_P66, color="g", ls="--")
ax[0].set_title("Daytime radiation with class thresholds")
ax[0].set_xlabel("W/mÂ²")

ax[1].hist(daily["Solar_kWh_m2"], bins=40)
ax[1].axvline(DAILY_P33, color="r", ls="--")
ax[1].axvline(DAILY_P66, color="g", ls="--")
ax[1].set_title("Daily solar energy with class thresholds")
ax[1].set_xlabel("kWh/mÂ² per day")
for a in ax:
    a.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

# Part 5 â€” Machine learning: predicting solar radiation
Target: `Solar_Radiation` (ALLSKY). Chronological split **70% train / 15% validation / 15% test**.
Three simple baselines first, then Linear Regression, Random Forest and Gradient Boosting.
Inputs: `Hour, Month, DayOfYear, Temperature, Rain, Clear_Sky_Solar_Radiation, Solar_Lag_24` (same hour, previous day).
Note: `Temperature`, `Rain` and `Clear_Sky` are for the *predicted hour* â€” in a real app they would come from a weather forecast (clear-sky radiation can be computed from date and location).

```python
# CELL 20 â€” Prepare X, y and Chronological Split
df["Solar_Lag_24"] = df["Solar_Radiation"].shift(24)       # safe: series is gapless (Cell 7)
ml = df.dropna(subset=["Solar_Lag_24"]).reset_index(drop=True)

features = ["Hour", "Month", "DayOfYear", "Temperature", "Rain",
            "Clear_Sky_Solar_Radiation", "Solar_Lag_24"]
target = "Solar_Radiation"

n = len(ml)
i_train = int(n * 0.70)
i_val = int(n * 0.85)

train, val, test = ml.iloc[:i_train], ml.iloc[i_train:i_val], ml.iloc[i_val:]

print("Features:", features)
print("\nTrain:", len(train), "rows |", train["time"].min().date(), "->", train["time"].max().date())
print("Val  :", len(val), "rows |", val["time"].min().date(), "->", val["time"].max().date())
print("Test :", len(test), "rows |", test["time"].min().date(), "->", test["time"].max().date())
```

```python
# CELL 21 â€” Baselines, Models and Comparison Table
def postprocess(pred, clear_sky):
    # Physical constraints: radiation cannot be negative, and is zero when the clear-sky value is zero
    pred = np.clip(np.asarray(pred, dtype=float), 0, None)
    return np.where(np.asarray(clear_sky) <= 0, 0.0, pred)

def metrics(y_true, y_pred):
    return {"MAE": mean_absolute_error(y_true, y_pred),
            "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
            "R2": r2_score(y_true, y_pred)}

# --- Baselines ---
climatology = train.groupby(["Month", "Hour"])["Solar_Radiation"].mean()
def climatology_pred(part):
    idx = pd.MultiIndex.from_frame(part[["Month", "Hour"]])
    return climatology.reindex(idx).values

predictions = {   # name -> (val_pred, test_pred)
    "Baseline: yesterday same hour": (val["Solar_Lag_24"].values, test["Solar_Lag_24"].values),
    "Baseline: month-hour average":  (climatology_pred(val), climatology_pred(test)),
    "Baseline: clear-sky value":     (val["Clear_Sky_Solar_Radiation"].values, test["Clear_Sky_Solar_Radiation"].values),
}

# --- Models (ordered from simplest to most complex) ---
models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(n_estimators=60, max_depth=14, min_samples_leaf=5,
                                           n_jobs=-1, random_state=RANDOM_STATE),
    "Gradient Boosting": HistGradientBoostingRegressor(max_iter=200, random_state=RANDOM_STATE),
}
fitted = {}
for name, mdl in models.items():
    mdl.fit(train[features], train[target])
    fitted[name] = mdl
    predictions[name] = (mdl.predict(val[features]), mdl.predict(test[features]))
print("Models trained.")
```

```python
# CELL 22 â€” Model Comparison Table
rows = []
test_day = test["Is_Daytime"].values
for name, (pv_, pt_) in predictions.items():
    pv_ = postprocess(pv_, val["Clear_Sky_Solar_Radiation"].values)
    pt_ = postprocess(pt_, test["Clear_Sky_Solar_Radiation"].values)
    predictions[name] = (pv_, pt_)
    mv = metrics(val[target], pv_)
    mt = metrics(test[target], pt_)
    md_ = metrics(test[target].values[test_day], pt_[test_day])
    rows.append({"Model": name,
                 "Val_MAE": mv["MAE"], "Val_RMSE": mv["RMSE"], "Val_R2": mv["R2"],
                 "Test_MAE": mt["MAE"], "Test_RMSE": mt["RMSE"], "Test_R2": mt["R2"],
                 "Test_Daytime_MAE": md_["MAE"], "Test_Daytime_RMSE": md_["RMSE"]})

comparison_table = pd.DataFrame(rows).set_index("Model").round(3)
display(comparison_table)
print("Units: MAE/RMSE in W/mÂ². 'Daytime' columns exclude night hours (night zeros make overall RÂ² look better).")
```

# Part 4 â€” PV generation estimate (MVP)
`ALLSKY_SFC_SW_DWN` is irradiance **G in W/mÂ²** (see units section). The standard rating condition of a PV panel is **1000 W/mÂ²**, so:

`PV_kW = PV_capacity_kW Ã— (G / 1000) Ã— [1 + Î³Â·(T_cell âˆ’ 25)] Ã— system_efficiency`, limited to the capacity (inverter limit)

* `T_cell = T_air + 0.03125 Ã— G` (simple NOCT-45 Â°C approximation)
* `Î³ = âˆ’0.004 per Â°C` (typical crystalline silicon)
* `system_efficiency` covers inverter, wiring, dirt and other losses (assumed)

Hourly energy: `kWh = kW Ã— 1 h`. This is an MVP estimate, not an engineering-grade simulation.

```python
# CELL 25 â€” User Inputs and PV Model
# ---------------- USER INPUTS ----------------
pv_capacity_kw = 3.0            # USER INPUT
system_efficiency = 0.80        # USER INPUT (inverter/wiring/dirt losses)
battery_capacity_kwh = 5.0      # USER INPUT
initial_soc_pct = 70.0          # USER INPUT

# ---------------- ASSUMPTIONS ----------------
GAMMA = -0.004                  # power temperature coefficient (per Â°C)
CELL_TEMP_RISE = 0.03125        # Â°C per W/m2 (NOCT 45 Â°C approx.)
BATT_EFF_CHARGE = 0.95          # ASSUMED
BATT_EFF_DISCHARGE = 0.95       # ASSUMED  (round trip ~ 90%)
BATT_MAX_RATE_C = 0.5           # ASSUMED: max charge/discharge power = 0.5 x capacity (kW)

def pv_power_kw(radiation_wm2, temp_c, capacity_kw=pv_capacity_kw, efficiency=system_efficiency):
    g = np.asarray(radiation_wm2, dtype=float)
    t_cell = np.asarray(temp_c, dtype=float) + CELL_TEMP_RISE * g
    power = capacity_kw * (g / 1000.0) * (1 + GAMMA * (t_cell - 25.0)) * efficiency
    return np.clip(power, 0, capacity_kw)

# Test-period table: real radiation -> PV_Actual ; ML-predicted radiation -> PV_Pred
res = test.copy()
res["Pred_Solar"] = pred_test
res["PV_Actual_kW"] = pv_power_kw(res["Solar_Radiation"], res["Temperature"])
res["PV_Pred_kW"] = pv_power_kw(res["Pred_Solar"], res["Temperature"])

print("PV capacity:", pv_capacity_kw, "kW | efficiency:", system_efficiency)
print("Peak estimated PV power in the test period:", round(res["PV_Actual_kW"].max(), 2), "kW")
res[["time", "Solar_Radiation", "Pred_Solar", "Temperature", "PV_Actual_kW", "PV_Pred_kW"]].iloc[8000:8008].round({"Solar_Radiation": 2, "Pred_Solar": 2, "Temperature": 2, "PV_Actual_kW": 2, "PV_Pred_kW": 2})
```

# Part 8 â€” Battery and energy-flow simulation
Hourly time step. Each hour:
1. `direct = min(PV, demand)` â€” solar used directly by the household.
2. Surplus PV charges the battery, limited by free capacity, the maximum charge rate and charging efficiency.
3. Whatever cannot be stored is **curtailed (unused)**.
4. If demand exceeds PV, the battery discharges (limited by stored energy, maximum rate, efficiency). Any remaining demand is **unmet by solar/battery** (it would come from the grid or a generator).

SOC always stays between 0% and 100%.

```python
# CELL 29 â€” Battery Simulation Function
def simulate(load, pv, cap_kwh, soc0_pct, eta_c=BATT_EFF_CHARGE, eta_d=BATT_EFF_DISCHARGE, rate_c=BATT_MAX_RATE_C):
    load = np.asarray(load, dtype=float).ravel()
    pv = np.asarray(pv, dtype=float).ravel()
    soc = cap_kwh * soc0_pct / 100.0
    rate = cap_kwh * rate_c                                  # kWh per hour
    out = {k: np.zeros(len(load)) for k in ["direct", "to_batt", "curtailed", "batt_out", "unmet", "soc_kwh"]}
    for t in range(len(load)):
        direct = min(pv[t], load[t])
        surplus, deficit = pv[t] - direct, load[t] - direct
        room = (cap_kwh - soc) / eta_c if cap_kwh > 0 else 0.0
        charge_in = min(surplus, rate, room)
        soc += charge_in * eta_c
        delivered = min(deficit, rate, soc * eta_d)
        soc -= delivered / eta_d
        out["direct"][t], out["to_batt"][t] = direct, charge_in
        out["curtailed"][t] = surplus - charge_in
        out["batt_out"][t], out["unmet"][t] = delivered, deficit - delivered
        out["soc_kwh"][t] = soc
    sim = pd.DataFrame(out)
    sim["pv"], sim["load"] = pv, load
    sim["soc_pct"] = sim["soc_kwh"] / cap_kwh * 100 if cap_kwh > 0 else 0.0
    assert sim["soc_pct"].between(-1e-9, 100 + 1e-9).all(), "SOC out of bounds"
    return sim

def summarize(sim):
    pv_total, demand = sim["pv"].sum(), sim["load"].sum()
    return {
        "Solar generation (kWh)": pv_total,
        "Household demand (kWh)": demand,
        "Solar used directly (kWh)": sim["direct"].sum(),
        "Solar sent to battery (kWh)": sim["to_batt"].sum(),
        "Solar curtailed/unused (kWh)": sim["curtailed"].sum(),
        "Battery discharge to household (kWh)": sim["batt_out"].sum(),
        "Unmet demand - grid/other (kWh)": sim["unmet"].sum(),
        "Direct Solar Utilization (%)": sim["direct"].sum() / pv_total * 100,
        "Demand met directly by solar (%)": sim["direct"].sum() / demand * 100,
        "Demand met by battery (%)": sim["batt_out"].sum() / demand * 100,
        "Demand unmet by solar+battery (%)": sim["unmet"].sum() / demand * 100,
    }

# Quick self-test: energy balance must close for every hour
_t = simulate(day_load(NORMAL_PROFILES[MAIN_PROFILE], True), pv_act[100], battery_capacity_kwh, initial_soc_pct)
assert np.allclose(_t["pv"], _t["direct"] + _t["to_batt"] + _t["curtailed"])
assert np.allclose(_t["load"], _t["direct"] + _t["batt_out"] + _t["unmet"])
print("Battery simulation self-test passed (PV and demand balances close, SOC within 0-100%).")
```

# Part 9 â€” Smart scheduling algorithm (transparent rules)
For every day, loads are placed in this order: fixed loads (Essential and non-schedulable) â†’ Important â†’ Flexible (largest energy first).
For each schedulable appliance every allowed start hour is scored by:

**score = energy of the appliance run that would be covered by the forecast solar surplus** (forecast PV minus the load already placed).

The highest score wins; ties go to the start time **closest to the usual time** (least disruption). If no start hour has any forecast solar surplus (very cloudy day), the appliance keeps its usual time.
Decisions use the **ML-forecast** PV; results are evaluated with the **real-radiation** PV.

```python
# CELL 30 â€” Scheduler
def schedule_day(pv_forecast, hot, normal_starts):
    starts = dict(normal_starts)
    active = [n for n, a in APPL.items() if (not a["hot"]) or hot]
    load = np.zeros(24)
    for n in active:                                         # fixed loads first
        a = APPL[n]
        if a["cat"] == "Essential" or not a["sched"]:
            load += run_profile(a["p"], a["dur"], starts[n])
    for group in ["Important", "Flexible"]:
        names = [n for n in active if APPL[n]["cat"] == group and APPL[n]["sched"]]
        names.sort(key=lambda n: -APPL[n]["p"] * APPL[n]["dur"])
        for n in names:
            a = APPL[n]
            best_key, best_start = None, a["normal"]
            for s in range(a["ws"], a["we"] + 1):
                if s + a["dur"] > 24:
                    continue
                L = run_profile(a["p"], a["dur"], s)
                surplus = np.clip(pv_forecast - load, 0, None)
                key = (round(float(np.minimum(L, surplus).sum()), 6), -abs(s - a["normal"]))
                if best_key is None or key > best_key:
                    best_key, best_start = key, s
            starts[n] = best_start if (best_key is not None and best_key[0] > 0) else a["normal"]
            load = load + run_profile(a["p"], a["dur"], starts[n])
    return starts, load

def build_loads(profile_name, pv_forecast_matrix=None):
    # Scenario A when pv_forecast_matrix is None, otherwise Scenario B
    normal = NORMAL_PROFILES[profile_name]
    loads, all_starts = [], []
    for i in range(len(dates)):
        if pv_forecast_matrix is None:
            loads.append(day_load(normal, hot_flags[i])); all_starts.append(normal)
        else:
            st, ld = schedule_day(pv_forecast_matrix[i], hot_flags[i], normal)
            loads.append(ld); all_starts.append(st)
    return np.array(loads), all_starts

# Example: one sunny summer day
example_i = int(np.argmax(np.where(hot_flags, pv_act.sum(axis=1), -1)))
st, _ = schedule_day(pv_pred[example_i], hot_flags[example_i], NORMAL_PROFILES[MAIN_PROFILE])
print("Example day:", dates[example_i].date())
print({k: f"{NORMAL_PROFILES[MAIN_PROFILE][k]:02d}:00 -> {v:02d}:00" for k, v in st.items()
       if v != NORMAL_PROFILES[MAIN_PROFILE][k]})
```

## Sensitivity: how much does the answer depend on our assumptions?
The size of the improvement depends on **when the household normally runs its appliances** and on **how large the PV system is** â€” both assumed. We therefore repeat the experiment for three usual-usage profiles and four PV capacities.

```python
# CELL 36 â€” Sensitivity Analysis
per_kw_act = pv_act / pv_capacity_kw
per_kw_pred = pv_pred / pv_capacity_kw

sens_rows = []
for prof in NORMAL_PROFILES:
    for cap in [1.0, 2.0, 3.0, 5.0]:
        pa, pp = per_kw_act * cap, per_kw_pred * cap
        sim_a = simulate(build_loads(prof)[0], pa, battery_capacity_kwh, initial_soc_pct)
        sim_b = simulate(build_loads(prof, pp)[0], pa, battery_capacity_kwh, initial_soc_pct)
        ua = sim_a["direct"].sum() / sim_a["pv"].sum() * 100
        ub = sim_b["direct"].sum() / sim_b["pv"].sum() * 100
        sens_rows.append({"Usual usage profile": prof, "PV (kW)": cap,
                          "Normal (%)": ua, "Smart (%)": ub, "Improvement (pp)": ub - ua})
sensitivity = pd.DataFrame(sens_rows)
display(sensitivity.round(1))

pivot = sensitivity.pivot(index="PV (kW)", columns="Usual usage profile", values="Improvement (pp)")
pivot.plot(kind="bar", figsize=(9, 4.5))
plt.title("Improvement in direct solar utilization (percentage points)")
plt.ylabel("pp")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
print("Range of improvement across all tested assumptions:",
      round(sensitivity["Improvement (pp)"].min(), 1), "to", round(sensitivity["Improvement (pp)"].max(), 1), "percentage points")
```

# Part 11 â€” Best-time recommendation
For a chosen appliance and day, every allowed start hour is evaluated using the **forecast** PV and the rest of the household's normal demand. For each window we report:
* expected solar availability class (from forecast radiation in the window)
* expected solar generation during the hours the run touches
* appliance energy, solar coverage of that energy, remaining surplus
* battery impact: energy that would have to come from the battery/grid, as % of battery capacity

Ranking rule (fully explainable): highest solar coverage â†’ highest expected solar generation â†’ closest to the usual time.

```python
# CELL 37 â€” Recommendation Function
def fmt_clock(x):
    return f"{int(x):02d}:{int(round((x % 1) * 60)):02d}"

def evaluate_window(name, start, i, pv_matrix, rad_matrix, normal_starts):
    a = APPL[name]
    base = day_load(normal_starts, hot_flags[i], exclude=name)
    pv = pv_matrix[i]
    L = run_profile(a["p"], a["dur"], start)
    touched = np.where(L > 0)[0]
    surplus = np.clip(pv - base, 0, None)
    covered = float(np.minimum(L, surplus).sum())
    energy = float(L.sum())
    draw = energy - covered
    return {
        "Window": f"{fmt_clock(start)}â€“{fmt_clock(start + a['dur'])}",
        "Solar_Availability": str(classify_hourly(rad_matrix[i][touched].mean())),
        "Expected_Solar_kWh": float(pv[touched].sum()),
        "Appliance_kWh": energy,
        "Solar_Coverage_%": covered / energy * 100,
        "Expected_Surplus_kWh": float(np.clip(surplus - L, 0, None)[touched].sum()),
        "Battery_Draw_kWh": draw,
        "Battery_Impact_%SOC": (-draw / battery_capacity_kwh * 100 + 0.0) if battery_capacity_kwh > 0 else 0.0,
        "_start": start,
    }

def recommend_time(name, i, normal_starts=NORMAL_PROFILES[MAIN_PROFILE], top_k=3):
    a = APPL[name]
    if not a["sched"]:
        raise ValueError(f"{name} is not schedulable (fixed load).")
    rows = []
    for s in range(a["ws"], a["we"] + 1):
        if s + a["dur"] <= 24:
            rows.append(evaluate_window(name, s, i, pv_pred, rad_pred, normal_starts))
    cand = pd.DataFrame(rows)
    cand["_closeness"] = -(cand["_start"] - normal_starts[name]).abs()
    cand = cand.sort_values(["Solar_Coverage_%", "Expected_Solar_kWh", "_closeness"], ascending=False).reset_index(drop=True)
    best = cand.iloc[0]
    usual = pd.DataFrame([evaluate_window(name, normal_starts[name], i, pv_pred, rad_pred, normal_starts)])
    # reality check with real radiation
    act_best = evaluate_window(name, int(best["_start"]), i, pv_act, rad_act, normal_starts)["Solar_Coverage_%"]
    act_usual = evaluate_window(name, normal_starts[name], i, pv_act, rad_act, normal_starts)["Solar_Coverage_%"]
    text = (f"Recommended: run the {name} at {best['Window']} (local time). Forecast solar availability is {best['Solar_Availability']}; "
            f"expected PV generation {best['Expected_Solar_kWh']:.2f} kWh covers {best['Solar_Coverage_%']:.0f}% of the appliance's "
            f"{best['Appliance_kWh']:.2f} kWh. Battery draw: {best['Battery_Draw_kWh']:.2f} kWh. "
            f"At the usual time ({usual.loc[0, 'Window']}) forecast coverage would be {usual.loc[0, 'Solar_Coverage_%']:.0f}%.")
    show = ["Window", "Solar_Availability", "Expected_Solar_kWh", "Appliance_kWh", "Solar_Coverage_%",
            "Expected_Surplus_kWh", "Battery_Draw_kWh", "Battery_Impact_%SOC"]
    return cand.head(top_k)[show], usual[show], text, (act_best, act_usual), cand

print("recommend_time() ready.")
```

# Limitations (please state these when presenting)
1. Household consumption and appliance usage times are **simulated assumptions**, not measured smart-meter data.
2. Appliance power ratings and durations are typical values and may differ from real devices (e.g. a refrigerator cycles on and off rather than running continuously).
3. PV generation is **estimated** from irradiance with a simple formula; no panel tilt, orientation, shading or inverter model.
4. The battery model is simplified (constant efficiency, fixed maximum power, no degradation, no depth-of-discharge limit).
5. The dataset covers **one location (Damascus)** and does not represent every Syrian household or region.
6. ML predictions depend on the historical dataset; the demo uses same-hour temperature, rain and clear-sky values as if they were forecasts. A live application needs an external weather forecast (clear-sky radiation can be computed from date and location).
7. Results are scenario-based; the improvement depends strongly on when the household normally uses its appliances and on PV size (see the sensitivity table). They are **not guaranteed real-world savings**.
8. The model works at whole-hour resolution (1.5 h runs start on the hour); the grid/unmet demand is not modelled in detail.
9. Timestamps are UTC (verified from the solar curve); the +3 h local offset is an explicit assumption.
